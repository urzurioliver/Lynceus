```python
import glob
import numpy as np

archivos = sorted(glob.glob("datos_crudos/20261103_120614_0001.npz"))   # los archivos de una corrida
partes = [np.load(f) for f in archivos]
d = {k: np.concatenate([p[k] for p in partes]) for k in partes[0].files}

n_iq = d["iq"].shape[1]
nombres_iq = [f"{'imag' if i % 2 == 0 else 'real'}{i // 2}" for i in range(n_iq)]
columnas = "t_s,t_pc,seq,rssi,ruido," + ",".join(nombres_iq)
tabla = np.column_stack([d["t_s"], d["t_pc"], d["seq"], d["rssi"], d["ruido"], d["iq"]])
np.savetxt("corrida.csv", tabla, delimiter=",", header=columnas, comments="",
           fmt=["%.6f", "%.3f"] + ["%d"] * (3 + n_iq))
```
